# 14. REBOUND N-body Simulations (Fast)

Gravitational dynamics: planetary systems, star clusters, BH binaries, GW captures.

```python
import rebound
import numpy as np
import matplotlib.pyplot as plt
np.random.seed(42)  # deterministic N-body draws
print('REBOUND', rebound.__version__)
```

## 14.1 Planetary system (fast)

```python
sim = rebound.Simulation()
sim.add(m=1.)
sim.add(m=1e-3, a=1., e=0.01)
sim.integrate(10.)  # Short integration
print('Position:', sim.particles[1].x, sim.particles[1].y)
```

## 14.2 Star cluster with BH binary (reduced N)

```python
sim = rebound.Simulation()
sim.add(m=1000.)
for i in range(20):  # Reduced from 50
    a = np.random.uniform(0.1, 5)
    sim.add(m=1., a=a, e=np.random.uniform(0, 0.15), inc=np.random.uniform(0, 0.25))
sim.add(m=10., a=1., e=0.1)
sim.add(m=10., a=1.2, e=0.1)
sim.integrate(100.)  # Reduced from 1000
print('Particles:', len(sim.particles))
```

## 14.3 GW capture cross-section

```python
def gw_capture_cross_section(m1, m2, v_inf):
    G = 1.0; c = 1.0
    M = m1 + m2
    eta = m1*m2/M**2
    sigma = 2*np.pi*(85*np.pi/12)**(2/7) * (G*M/c**2)**2 * (c/v_inf)**(18/7) * eta**(2/7)
    return sigma

for v in [10, 50, 100, 200]:
    print(f'v_inf={v}: sigma={gw_capture_cross_section(30, 30, v):.3e}')
```
