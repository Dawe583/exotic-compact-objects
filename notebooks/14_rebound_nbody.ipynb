# 14. REBOUND N-body Simulations

Gravitational dynamics: planetary systems, star clusters, BH binaries, GW captures.

```
import rebound
import numpy as np
import matplotlib.pyplot as plt
print('REBOUND', rebound.__version__)
```

## 14.1 Planetary system

```
sim = rebound.Simulation()
sim.add(m=1.)  # Sun
sim.add(m=1e-3, a=1., e=0.01)  # Earth-like
sim.add(m=1e-3, a=1.5, e=0.05)  # Mars-like
sim.integrate(100.)
print('Final positions:', [(p.x, p.y) for p in sim.particles[1:]])
```

## 14.2 Star cluster with BH binary

```
sim = rebound.Simulation()
sim.add(m=1000.)  # Central IMBH
for i in range(50):
    a = np.random.uniform(0.1, 10)
    sim.add(m=1., a=a, e=np.random.uniform(0, 0.5), inc=np.random.uniform(0, np.pi))
sim.add(m=10., a=1., e=0.1)  # BH binary primary
sim.add(m=10., a=1.2, e=0.1)  # BH binary secondary
sim.integrate(1000.)
print('Particles:', len(sim.particles))
```

## 14.3 GW capture cross-section

```
def gw_capture_cross_section(m1, m2, v_inf):
    # Approximate cross-section for GW capture
    G = 1.0  # code units
    c = 1.0
    M = m1 + m2
    eta = m1*m2/M**2
    # Peters 1964 formula
    sigma = 2*np.pi*(85*np.pi/12)**(2/7) * (G*M/c**2)**2 * (c/v_inf)**(18/7) * eta**(2/7)
    return sigma

for v in [10, 50, 100, 200]:
    print(f'v_inf={v} km/s: sigma={gw_capture_cross_section(30, 30, v):.3e}')
```
