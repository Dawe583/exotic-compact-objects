# 15. EinsteinPy Geodesics (Fast)

## 15.1 Timelike geodesic (short)

```
geod = Timelike(metric='Schwarzschild', metric_params=(), position=[10., np.pi/2, 0.], momentum=[0., 0., 3.5], steps=500, delta=0.5)
geod.calculate_trajectory()
traj = geod.trajectory[1]
r, phi = traj[:,1], traj[:,3]
plt.figure(figsize=(6,6))
plt.plot(r*np.cos(phi), r*np.sin(phi), lw=0.5)
th = np.linspace(0,2*np.pi,200)
plt.plot(2*np.cos(th), 2*np.sin(th), 'k-')
plt.gca().set_aspect('equal'); plt.title('Bound orbit'); plt.show()
```

## 15.2 Null geodesic - photon ring (short)

```
geo = Nulllike(metric='Schwarzschild', metric_params=(), position=[8., np.pi/2, 0.], momentum=[0., -0.1, 2.8], steps=1000, delta=0.2)
geo.calculate_trajectory()
traj = geo.trajectory[1]
r, phi = traj[:,1], traj[:,3]
plt.figure(figsize=(6,6))
plt.plot(r*np.cos(phi), r*np.sin(phi), lw=0.5)
plt.plot(3*np.cos(th), 3*np.sin(th), 'r--')
plt.gca().set_aspect('equal'); plt.title('Photon trajectory'); plt.show()
```
