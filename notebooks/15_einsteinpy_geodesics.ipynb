# 15. EinsteinPy Geodesics & Ray Tracing

Timelike/null geodesics for Schwarzschild, Kerr, Kerr-Newman. Shadows, ray tracing.

```
import numpy as np
import matplotlib.pyplot as plt
from einsteinpy.geodesic import Timelike, Null
from einsteinpy.metric import Schwarzschild, Kerr, KerrNewman
print('EinsteinPy OK')
```

## 15.1 Timelike geodesics (Schwarzschild)

```
geod = Timelike(metric='Schwarzschild', metric_params=(), position=[10., np.pi/2, 0.], momentum=[0., 0., 3.5], steps=2000, delta=1.0)
geod.calculate_trajectory()
traj = geod.trajectory[1]
r, phi = traj[:,1], traj[:,3]
plt.figure(figsize=(6,6))
plt.plot(r*np.cos(phi), r*np.sin(phi))
th = np.linspace(0,2*np.pi,200)
plt.plot(2*np.cos(th), 2*np.sin(th), 'k-')
plt.gca().set_aspect('equal'); plt.title('Bound orbit'); plt.show()
```

## 15.2 Null geodesics - photon ring

```
geo = Null(metric='Schwarzschild', metric_params=(), position=[10., np.pi/2, 0.], momentum=[0., -0.1, 3.], steps=5000, delta=0.5)
geo.calculate_trajectory()
traj = geo.trajectory[1]
r, phi = traj[:,1], traj[:,3]
plt.figure(figsize=(6,6))
plt.plot(r*np.cos(phi), r*np.sin(phi), lw=0.5)
plt.plot(3*np.cos(th), 3*np.sin(th), 'r--', label='Photon sphere')
plt.gca().set_aspect('equal'); plt.legend(); plt.title('Photon trajectory'); plt.show()
```

## 15.3 Kerr shadow

```
a = 0.9
geo = Null(metric='Kerr', metric_params=(a,), position=[20., np.pi/2, 0.], momentum=[0., -0.05, 4.5], steps=10000, delta=0.5)
geo.calculate_trajectory()
traj = geo.trajectory[1]
r, phi = traj[:,1], traj[:,3]
plt.figure(figsize=(6,6))
plt.plot(r*np.cos(phi), r*np.sin(phi), lw=0.5)
rplus = 1+np.sqrt(1-a**2)
plt.plot(rplus*np.cos(th), rplus*np.sin(th), 'k-')
plt.gca().set_aspect('equal'); plt.title(f'Kerr a={a} null geodesic'); plt.show()
```

## 15.4 Kerr-Newman (charged rotating)

```
geo = Null(metric='KerrNewman', metric_params=(0.9, 0.1), position=[20., np.pi/2, 0.], momentum=[0., -0.05, 4.5], steps=10000, delta=0.5)
geo.calculate_trajectory()
traj = geo.trajectory[1]
r, phi = traj[:,1], traj[:,3]
plt.figure(figsize=(6,6))
plt.plot(r*np.cos(phi), r*np.sin(phi), lw=0.5)
plt.gca().set_aspect('equal'); plt.title('Kerr-Newman a=0.9, Q=0.1'); plt.show()
```
