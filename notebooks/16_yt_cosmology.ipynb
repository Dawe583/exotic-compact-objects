## 16.1 Small synthetic dataset

```
import yt
import numpy as np
N = 32
density = 1 + np.random.normal(0, 0.1, (N,N,N))
ds = yt.load_uniform_grid({'density': density}, (N,N,N), length_unit='Mpc')
print('Dataset:', ds.domain_dimensions)
```

## 16.2 Halo finding

```
try:
    from yt.analysis_modules.halo_finding.api import HaloFinder
    hf = HaloFinder(ds)
    print(f'Found {len(hf.halo_list)} halos')
except:
    print('No halos or halo finder not available')
```

## 16.3 Power spectrum

```
try:
    from yt.analysis_modules.power_spectrum.api import PowerSpectrum
    ps = PowerSpectrum(ds, fields=[('gas', 'density')])
    k_vals, pk_vals = ps.compute()
    import matplotlib.pyplot as plt
    plt.loglog(k_vals, pk_vals)
    plt.xlabel('k'); plt.ylabel('P(k)'); plt.title('Power spectrum')
    plt.grid(True, alpha=0.3); plt.show()
except:
    print('Power spectrum computation skipped')
```
