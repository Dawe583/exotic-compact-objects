## 16.1 Create synthetic cosmological dataset

```
import yt
from yt.frontends.stream.api import load_uniform_grid
import numpy as np

# Density field with power spectrum
N = 128
k = np.fft.fftfreq(N)*N
kx, ky, kz = np.meshgrid(k, k, k, indexing='ij')
k_mag = np.sqrt(kx**2 + ky**2 + kz**2)
Pk = np.where(k_mag>0, k_mag**(-3), 0)
phases = np.random.uniform(0, 2*np.pi, (N,N,N))
delta_k = np.sqrt(Pk) * np.exp(1j*phases)
delta = np.real(np.fft.ifftn(delta_k))
density = 1 + delta

bbox = np.array([[-0.5, 0.5]]*3)
ds = load_uniform_grid({'density': density}, density.shape, length_unit='Mpc', bbox=bbox, nprocs=1)
print('Dataset created:', ds.domain_dimensions)
```

## 16.2 Halo finding

```
from yt.analysis_modules.halo_finding.api import HaloFinder
hf = HaloFinder(ds)
halos = hf.halo_list
print(f'Found {len(halos)} halos')
if len(halos) > 0:
    for h in halos[:5]:
        print(f'  M={h.mass:.3e}, pos={h.position}')
```

## 16.3 Power spectrum

```
from yt.analysis_modules.power_spectrum.api import PowerSpectrum
ps = PowerSpectrum(ds, fields=[('gas', 'density')])
k_vals, pk_vals = ps.compute()
plt.loglog(k_vals, pk_vals)
plt.xlabel('k (Mpc^-1)'); plt.ylabel('P(k)'); plt.title('Matter power spectrum')
plt.grid(True, alpha=0.3); plt.show()
```
