```
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.ensemble import RandomForestRegressor, RandomForestClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error, accuracy_score
import xgboost as xgb
print('ML libs OK')
```

## 18.1 BH mass prediction from features

```
np.random.seed(42)
n = 10000
# Features: Lx, T, variability, redshift, Fe_K_alpha_EW
Lx = np.random.lognormal(42, 1, n)
T = np.random.lognormal(7, 0.5, n)
var = np.random.exponential(0.3, n)
z = np.random.exponential(0.5, n)
fe_ew = np.random.exponential(100, n)
# True mass: M ~ Lx^0.5 * T^0.25
M_true = 1e6 * (Lx/1e42)**0.5 * (T/1e7)**0.25 * np.random.lognormal(0, 0.1, n)
X = np.column_stack([np.log10(Lx), np.log10(T), var, z, np.log10(fe_ew+1)])
y = np.log10(M_true)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)
model = RandomForestRegressor(n_estimators=200, max_depth=15)
model.fit(X_train, y_train)
y_pred = model.predict(X_test)
print(f'RMSE: {np.sqrt(mean_squared_error(y_test, y_pred)):.4f} dex')
plt.scatter(y_test, y_pred, alpha=0.3, s=1)
plt.plot([6, 10], [6, 10], 'r--')
plt.xlabel('True log M'); plt.ylabel('Pred log M'); plt.title('BH mass prediction'); plt.show()
```

## 18.3 BH/NS/WH classifier

```
np.random.seed(42)
n = 5000
# Features: mass, spin, tidal_deformability, GW_freq, EM_counterpart
mass = np.concatenate([np.random.lognormal(1, 0.5, n//3), np.random.lognormal(0.3, 0.3, n//3), np.random.lognormal(3, 1, n//3)])
spin = np.concatenate([np.random.uniform(0, 0.99, n//3), np.random.uniform(0, 0.4, n//3), np.random.uniform(0, 0.99, n//3)])
tidal = np.concatenate([np.random.exponential(10, n//3), np.random.uniform(0, 5000, n//3), np.random.exponential(1e6, n//3)])
labels = np.array(['BH']*(n//3) + ['NS']*(n//3) + ['WH']*(n//3))
X = np.column_stack([np.log10(mass), spin, np.log10(tidal+1)])
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
y = le.fit_transform(labels)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)
clf = RandomForestClassifier(n_estimators=200)
clf.fit(X_train, y_train)
print(f'Accuracy: {accuracy_score(y_test, clf.predict(X_test)):.3f}')
print('Classes:', le.classes_)
```
