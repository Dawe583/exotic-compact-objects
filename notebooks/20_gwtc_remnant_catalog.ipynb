# GWTC: katalog 20 nedávných pozůstatků mergerů černých děr

Výběr obsahuje posledních 20 jedinečných událostí s uvedenou konečnou hmotností v lokálním GWTC-4.1/GWTC-5.0 katalogu. Pozor: jde o **pozůstatky splynutí**, nikoli 20 izolovaných černých děr. Nejde o reprezentativní výběr pro populační závěry.

```python
from pathlib import Path
import pandas as pd
import plotly.express as px
root=Path.cwd()
if not (root/'data/processed/gwtc_recent_20_remnants.csv').exists(): root=Path('../')
df=pd.read_csv(root/'data/processed/gwtc_recent_20_remnants.csv')
df.head()
```

```python
# Notebook metadata
__author__ = "Astro Science Research"
__date__ = "2026-10-28"
__version__ = "1.0.0"
__kernel__ = "python3"
import sys
print(f"Python {sys.version}")
print(f"Notebook: {__file__ if '__file__' in dir() else 'interactive'}")
```

```python
fig=px.scatter(df,x='mass_1_source_msun',y='mass_2_source_msun',color='final_mass_source_msun',size='network_snr',hover_name='event_id',hover_data=['catalog','luminosity_distance_mpc','redshift','chi_eff','p_astro','source_url'],color_continuous_scale='Plasma',template='plotly_dark',title='Komponentové hmotnosti a hmotnost pozůstatku')
fig.update_layout(xaxis_title='Primární zdrojová hmotnost (M☉)',yaxis_title='Sekundární zdrojová hmotnost (M☉)')
fig.show()
```

```python
fig=px.scatter(df,x='luminosity_distance_mpc',y='final_mass_source_msun',color='chi_eff',hover_name='event_id',hover_data=['mass_1_source_msun','mass_2_source_msun','p_astro','source_url'],log_x=True,color_continuous_scale='RdBu',template='plotly_dark',title='Hmotnost pozůstatku, vzdálenost a efektivní spin')
fig.update_layout(xaxis_title='Luminosity distance (Mpc)',yaxis_title='Final source-frame mass (M☉)')
fig.show()
```
