# AGN supermassive black-hole mass catalog

This notebook analyzes 70 numeric entries from the local 86-row AGN reverberation-mapping compilation. Values are method-dependent literature measurements, not a homogeneous or completeness-corrected survey. Uncertainties are supplied in log10 mass and converted to linear asymmetric mass offsets.

```python
from pathlib import Path
import pandas as pd, numpy as np
import plotly.express as px
root=Path.cwd()
if not (root/'data/processed/agn_reverberation_masses.csv').exists(): root=Path('../')
d=pd.read_csv(root/'data/processed/agn_reverberation_masses.csv').dropna(subset=['log10_mass_msun'])
d[['name','redshift','log10_mass_msun','mass_msun','mass_lower_delta_msun','mass_upper_delta_msun']].head()
```

```python
fig=px.scatter(d,x='redshift',y='log10_mass_msun',color='log10_mass_msun',hover_name='name',hover_data=['mass_msun','logmass_lower_error_dex','logmass_upper_error_dex'],color_continuous_scale='Plasma',template='plotly_dark',title='AGN reverberation masses by redshift')
fig.update_layout(xaxis_title='Redshift z',yaxis_title='log₁₀(M/M☉)')
fig.show()
```

```python
# Descriptive, not population-inference statistics; literature selection effects are unmodeled.
d.log10_mass_msun.describe(percentiles=[.1,.5,.9])
```
